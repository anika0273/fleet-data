```python
import psycopg2

conn = psycopg2.connect(
    host="postgres",    # service name in docker-compose
    port=5432,
    database="fleet_db",
    user="postgres",
    password="1234"
)
print("Connected to Postgres")
```

```python
import pandas as pd
from sqlalchemy import create_engine, inspect

# Adjust these settings to match your Docker config
engine = create_engine("postgresql://postgres:1234@localhost:5432/fleet_db")
inspector = inspect(engine)
table_names = inspector.get_table_names()
print(table_names)
```
